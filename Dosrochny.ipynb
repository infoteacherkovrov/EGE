```python
# 5
def f(n):
    s = ''
    while n > 0:
        s = str(n % 2) + s
        n = n // 2
    return s


for n in range(1, 1000):
    b = f(n)
    for j in range(2):
        sumdigits = sum(map(int, b))
        b = b + str(sumdigits % 2)
    r = int(b, 2)
    if r > 253:
        print(n)
        break
```

```python
#16 время выполнения около 2 минут
from functools import lru_cache
@lru_cache(None)
def f(n):
    if n < 10:
        return 3
    else:
        return (n + 4) * f(n - 5)
for i in range(10,257490):
    f(i)
print((f(257487)//683 + 67 * f(257477))//f(257472))
```

```python
#23
def f(n, c):
    if n == 7 or n > 25:
        return False
    if n == 15:
        c += 1
    if n == 25:
        if c == 1:
            return 1
        else:
            return 0
    return f(n + 1, c) + f(n + 3, c) + f(n * 2, c)
print(f(2,0))
```
